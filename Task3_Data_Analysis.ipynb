```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/train.csv")

df.head()
```

```python
# Show basic information about dataset (columns, missing values, data types)
df.info()
```

```python
# Show statistical summary of numerical columns
df.describe()
```

```python
# Fill missing values in 'Age' column using median (robust to outliers)
df["Age"].fillna(df["Age"].median(), inplace=True)

# Fill missing values in 'Embarked' column using mode (most frequent value)
df["Embarked"].fillna(df["Embarked"].mode()[0], inplace=True)

# Drop 'Cabin' column because it has too many missing values
df.drop("Cabin", axis=1, inplace=True)

# Remove duplicate rows if any exist
df.drop_duplicates(inplace=True)

# Verify that missing values are handled
df.info()
```

```python
# Calculate survival rate for each gender
# Mean works because Survived = 1 (alive), 0 (dead)
survival_by_gender = df.groupby("Sex")["Survived"].mean()

survival_by_gender
```

```python
# Analyze survival rate based on passenger class
survival_by_class = df.groupby("Pclass")["Survived"].mean()

survival_by_class
```

```python
# Calculate average age for each passenger class
average_age_per_class = df.groupby("Pclass")["Age"].mean()

average_age_per_class
```

```python
# Create age groups using pd.cut (feature engineering)
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 12, 18, 60, 100],
    labels=["Child", "Teen", "Adult", "Senior"]
)

# Calculate survival rate for each age group
survival_by_age_group = df.groupby("AgeGroup")["Survived"].mean()

survival_by_age_group
```

```python
# Filter female passengers who survived
female_survivors = df[(df["Sex"] == "female") & (df["Survived"] == 1)]

female_survivors.head()
```

```python
# Filter children (Age < 12) who survived
children_survivors = df[(df["Age"] < 12) & (df["Survived"] == 1)]

children_survivors.head()
```

```python
# Filter passengers in 1st class who survived
first_class_survivors = df[(df["Pclass"] == 1) & (df["Survived"] == 1)]

first_class_survivors.head()
```

📊 Insights

1. Gender:
Females were much more likely to survive than males. This suggests that women were given priority during evacuation.

2. Passenger Class:
First-class passengers had a higher survival rate compared to second and third class. This shows that social class had a strong impact on survival chances.

3. Children:
Children had a relatively higher survival rate, indicating that rescue efforts prioritized younger passengers.

4. Highest Survival Group:
The group with the highest survival probability was female passengers in first class.

5. Conclusion:
This analysis shows that survival was not random. It was influenced by gender, class, and age. Therefore, analyzing patterns in data helps us understand real-world decision-making.
